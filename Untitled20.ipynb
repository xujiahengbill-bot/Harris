```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
```

```python
# Load data
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
```

```python
def preprocess(df):
    df = df.copy()

    # Fill missing values
    df["Age"] = df["Age"].fillna(df["Age"].median())
    df["Fare"] = df["Fare"].fillna(df["Fare"].median())
    df["Embarked"] = df["Embarked"].fillna("S")

    # Encode categorical variables
    df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
    df["Embarked"] = df["Embarked"].map({"S": 0, "C": 1, "Q": 2})

    # Family features
    df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
    df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

    # Select final features
    features = [
        "Pclass", "Sex", "Age", "Fare",
        "Embarked", "FamilySize", "IsAlone"
    ]

    return df[features]
```

```python
X = preprocess(train)
y = train["Survived"]

X_test = preprocess(test)
```

```python
log_model = Pipeline([
    ("scaler", StandardScaler()),
    ("logreg", LogisticRegression(max_iter=1000))
])

log_score = cross_val_score(log_model, X, y, cv=5, scoring="accuracy")
print("Logistic Regression CV Accuracy:", log_score.mean())
```

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=6,
    random_state=42
)

rf_score = cross_val_score(rf_model, X, y, cv=5, scoring="accuracy")
print("Random Forest CV Accuracy:", rf_score.mean())
```

```python
gb_model = GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

gb_score = cross_val_score(gb_model, X, y, cv=5, scoring="accuracy")
print("Gradient Boosting CV Accuracy:", gb_score.mean())
```

```python
from sklearn.ensemble import VotingClassifier

log_model = Pipeline([
    ("scaler", StandardScaler()),
    ("logreg", LogisticRegression(max_iter=1000))
])

rf_model = RandomForestClassifier(
    n_estimators=200, max_depth=6, random_state=42
)

gb_model = GradientBoostingClassifier(
    n_estimators=200, learning_rate=0.05, max_depth=3, random_state=42
)

voting_model = VotingClassifier(
    estimators=[
        ("log", log_model),
        ("rf", rf_model),
        ("gb", gb_model)
    ],
    voting="soft"
)

voting_score = cross_val_score(voting_model, X, y, cv=5, scoring="accuracy")
print("Soft Voting CV Accuracy:", voting_score.mean())
```

```python
weighted_voting = VotingClassifier(
    estimators=[
        ("log", log_model),
        ("rf", rf_model),
        ("gb", gb_model)
    ],
    voting="soft",
    weights=[1, 2, 3]
)

weighted_score = cross_val_score(weighted_voting, X, y, cv=5, scoring="accuracy")
print("Weighted Voting CV Accuracy:", weighted_score.mean())
```

```python
from sklearn.ensemble import StackingClassifier
stack_model = StackingClassifier(
    estimators=[
        ("log", log_model),
        ("rf", rf_model),
        ("gb", gb_model)
    ],
    final_estimator=LogisticRegression(),
    cv=5,
    passthrough=False
)

stack_score = cross_val_score(stack_model, X, y, cv=5, scoring="accuracy")
print("Stacking CV Accuracy:", stack_score.mean())
```

```python
from xgboost import XGBClassifier
xgb_model = XGBClassifier(
    n_estimators=600,
    learning_rate=0.03,
    max_depth=3,
    subsample=0.9,
    colsample_bytree=0.9,
    reg_lambda=1.0,
    reg_alpha=0.0,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)
```

```python
stack_model = StackingClassifier(
    estimators=[
        ("log", log_model),
        ("rf", rf_model),
        ("gb", gb_model),
        ("xgb", xgb_model)
    ],
    final_estimator=LogisticRegression(max_iter=2000),
    cv=5,
    stack_method="predict_proba",  # important for classification
    passthrough=False
)

stack_score = cross_val_score(stack_model, X, y, cv=5, scoring="accuracy")
print("Stacking (with XGBoost) CV Accuracy:", stack_score.mean())
stack_model.fit(X, y)

preds = stack_model.predict(X_test)

submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": preds
})

submission.to_csv("stack_xgb_submission.csv", index=False)
print("Saved: stack_xgb_submission.csv")
```
